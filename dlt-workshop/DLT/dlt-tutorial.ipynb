## simulating how dlt works using simple dictionary

```python
import dlt
```

```python

@dlt.resource
def data():
    yield {"name": "Alice"}
    yield {"name": "Bob"}

pipeline = dlt.pipeline(
    pipeline_name="tutor_pipeline",
    destination="duckdb",
    dataset_name="data"
)

pipeline.run(data())
```

```python
This creates data_set name called `data`. You can visualize it using duckdb interface {duckdb -ui)
```

### Stages of dlt

```python
Dlt has three(3} distinct stages 

1. **Extract**: download raw data from the API  
2. **Normalize**: turn nested JSON into relational tables  
3. **Load**: write those tables into DuckDB 




```
